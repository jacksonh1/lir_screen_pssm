```python
import re

pattern = re.compile(r"(?=([FWY]..[LVIFWY]))")  # lookahead captures overlapping matches

def extract_7mers(seq):
    matches = []
    found = re.findall(pattern, seq)  # gives list of matching substrings

    # walk through the sequence and look for each match's position
    for motif in found:
        # find the start position of the motif (could be multiple identical motifs, so loop!)
        start_idx = 0
        while True:
            start = seq.find(motif, start_idx)
            if start == -1:
                break
            if start >= 3:
                seven_mer = seq[start - 3 : start + 4]
                matches.append(seven_mer)
            start_idx = start + 1  # allow overlapping matches
    return matches


df4['ID2'] = 'PLR' + df4['ID']
df4['7mers'] = df4['ID2'].apply(extract_7mers)
df4 = df4[df4['7mers'].str.len() > 0]
df_7mers = df4.explode('7mers').reset_index(drop=True)
```

```python
import re

pattern = r'[FWY]..[LVIFWY]'

def extract_7mers(seq):
    matches = []
    for match in re.finditer(pattern, seq):
        start = match.start()
        if start >= 3:  # Ensure there are 3 residues before the motif
            seven_mer = seq[start-3:start+4]
            matches.append(seven_mer)
    return matches
df8['ID2'] = 'PLR' + df8['ID']
# Apply to the DataFrame
df8['7mers'] = df8['ID2'].apply(extract_7mers)

# Drop rows with no 7-mers
df8 = df8[df8['7mers'].str.len() > 0]

# Explode so each 7-mer is its own row
df_7mers2 = df8.explode('7mers').reset_index(drop=True)
```
