```python
import pandas as pd
import lir_proteome_screen_pssm.environment as env
```

```python
import lir_proteome_screen_pssm.data_loaders as dl

pst = dl.get_processed_sequence_tables()
pst.ilir_binders
```

```python
lc = pd.read_csv(env.RAW_DATA_DIR / "LIRcentral - Data export.csv")
```

```python
lc
```

```python
lc["UNIPROT ACC"].isin(pst.ilir_binders["_UNIPROT ACC"]).sum()
```

```python

lc["Up-stream"] = lc["Up-stream"].fillna("")
lc["Down-stream"] = lc["Down-stream"].fillna("")
lc["Motif"] = lc["Motif"].fillna("")
lc["motif_start_in_hit"] = lc["Up-stream"].str.len()
lc["len_downstream"] = lc["Down-stream"].str.len()
lc["len_motif"] = lc["Motif"].str.len()


def build_7mer(row: pd.Series):
    return row["Up-stream"][-3:] + row["Motif"]


lc["full_seq"] = lc.apply(build_7mer, axis=1)
```

```python
lc['full_seq'].isin(pst.ilir_binders['7mer']).sum()
```
